# Problem Solving 103: Detective Work with Descriptive Stats

### Step 1 - Warm Up on Mean vs. Median as a Signal

#1. Mean noticeably higher than median

##### When the mean is noticeably higher than the median, it means that the data is right-skewed, meaning a few unusually large values are pulling the average upward. Therefore; if the mean is noticeably lower, it means the data is left-skewed. 

#3. Why can the mean and median legitimately be far apart?
 ##### A store may occasionally receive very large wholesale orders, creating a few unusually high sales days.

### Step 2 - Learn to Compare Consistency, Not Just Averages

##### Standard deviation tells us how spread out the sales are around the mean. A smaller standard deviation generally means more consistent sales.

#Step 6 — Limits of Summary Statistics

##### The mean, median, and standard deviation cannot tell me exactly which days caused the unusual pattern or whether the large sales values are legitimate. I would need to see the raw data, a histogram, and additional business context to confirm what is happening.

##### This activity prepares me for inferential statistics because it teaches me to distinguish between a pattern that appears in the data and a pattern that can be shown to be statistically meaningful rather than simply caused by random variation.
