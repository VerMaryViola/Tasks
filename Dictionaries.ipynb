## Dictionaries: Structured data

```python
Products = [
    
     {'name': "Tomato Paste (tin)", 'price': 378.0, 'quantity': 16.0 },
     {'name': "Bread (loaf)", 'price': 591.0, 'quantity': 13.0},
     {'name': "Salt 1kg", 'price': 297.0, 'quantity': 11.0},
     {'name': "Rice 50kg", 'price': 34928.0, 'quantity': 1.0},
     {'name': "Detergent 1kg", 'price': 1768.0, 'quantity': 10.0}    
]
```

```python
for product in Products:
    print(product)
```

```python
print("NKWEN TRADERS PRODUCT INVENTORY REPORT")
for product in Products:
    print(f"Name: {product['name']}")
    print(f"Price: {product['price']}")
    print(f"Quantity: {product['quantity']}")
    print()
```

```python
Total = 0
for product in Products:
    inventory_value = product['price'] * product['quantity']
    Total += inventory_value

    
```

```python
print(f"Total inventory value: {Total} FCFA")
```
