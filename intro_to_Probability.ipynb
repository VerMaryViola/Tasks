# Intro to Probability

## Problem 1 :  Learn the 0–1 Scale

##### - 0: The event is impossible; it cannot happen.
##### - 0.5: The event has an equal chance of happening or not happening.
##### - 1: The event is certain; it must happen.

##### Examples:
##### - 0: A customer buys a product that does not exist in the shop.
##### - 0.5: Getting heads when flipping a fair coin.
##### - 1: A week has 7 days.

## Step 2 - Learn Independent vs. Dependent EventsIndependent Event Pair

##### Example: Tossing a fair coin twice.

##### Explanation: The result of the first coin toss does not affect the result of the second toss, so the two events are independent.

##### Dependent Event Pair

##### Example: Taking two sweets from a bag without replacing the first sweet.

##### Explanation: The first sweet taken changes what remains in the bag, so it affects the probability of getting a particular sweet on the second draw; therefore, the events are dependent.

## Step 3 — Conditional Probability Intuition

##### Conditional probability means looking only at the smaller group that already meets a condition, and then finding what fraction of that group also has the outcome we are interested in.
##### For example:
##### If 3 out of 10 customers who bought rice also bought sugar, we ignore customers who did not buy rice. We only look at the 10 rice buyers, so the probability is 3 out of 10, or 0.3.


## Step 4 - Solve the Practice Problems
##### Problem 1 - Rice and Sugar
##### Type: Conditional probability

```python
rice_buyers = 10
rice_and_sugar = 3

probability = rice_and_sugar / rice_buyers

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

##### Problem 2 : Store Visits
##### Type: Conditional probability
##### There are 200 customers, and 50 of them are rice buyers meaning that the probability is 50 out of 200.

```python
total_customers = 200
rice_buyers = 50

probability = rice_buyers / total_customers

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

##### Problem 4 - Bread and Milk 
##### Type: It is an independent check

```python
bread = 0.30
bread_given_milk = 0.30

print("Overall probability of bread:", bread)
print("Probability of bread among milk buyers:", bread_given_milk)

if bread == bread_given_milk:
    print("The events are independent.")
else:
    print("The events are dependent.")
```

##### Independent because, knowing that a customer bought milk does not change the probability of buying bread and it is 0.30 in both cases.

##### Problem 5 - Rice Purchase and Late Delivery 
##### Type: This is an independent check

```python
prob_rice = 0.25
prob_late_delivery = 0.10

probability = prob_rice * prob_late_delivery

print("Probability:", probability)
print("Percentage:", probability * 100, "%")
```

##### Problem 6 - Sugar and Cooking Oil (One Week) 
##### Type: It is a dependent check

```python
cooking_oil = 25
total_customers = 120

probability = cooking_oil / total_customers

print("(a)")
print("Probability:", probability)
print(f"{probability:.2f}%")  # to print in two decimal places
```

```python
sugar_buyers = 80
bought_cooking_oil = 20
prob_cooking_oil_given_bought_sugar = bought_cooking_oil / sugar_buyers

print("(b)")
print("Probability(cooking oil | bought sugar)", prob_cooking_oil_given_bought_sugar)
print(f"{prob_cooking_oil_given_bought_sugar * 100:.2f}%")
```

```python
print("(c)")
if cooking_oil == prob_cooking_oil_given_bought_sugar:
    print("It is independent.")
else:
    print("It is dependent.")
    
```
