## Snippet 4:

```python
products = [
    {"name": "Rice 25kg Bag", "price": 15000, "quantity": 12},
    {"name": "Sugar 1kg", "price": 900, "quantity": 50},
    {"name": "Bread Loaf", "price": 700, "quantity": 30},
]
total_value = 0
for product in products:   
    total_value = product["price"] * product["quantity"]
print(f"Total inventory value: {total_value} FCFA")
```

```python
products = [
    {"name": "Rice 25kg Bag", "price": 15000, "quantity": 12},
    {"name": "Sugar 1kg", "price": 900, "quantity": 50},
    {"name": "Bread Loaf", "price": 700, "quantity": 30},
]
total_value = 0
for product in products:   
    total_value += product["price"] * product["quantity"]
print(f"Total inventory value: {total_value} FCFA")
```

#### Problem:
##### The code kept replcing the total value of the next instead of summing it all.
#### Solution: 
##### I just replced the = with += so it sums it to produce the total of all.
