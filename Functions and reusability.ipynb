## Functions & Reusable Logic

```python
def product_dict():
    Products = [
     {'name': "Tomato Paste (tin)", 'price': 378.0, 'quantity': 16.0 },
     {'name': "Bread (loaf)", 'price': 591.0, 'quantity': 13.0},
     {'name': "Salt 1kg", 'price': 297.0, 'quantity': 11.0},
     {'name': "Rice 50kg", 'price': 34928.0, 'quantity': 1.0},
     {'name': "Detergent 1kg", 'price': 1768.0, 'quantity': 10.0}    
 ] 
    return Products
product_dict()
    
```

```python
Products = product_dict()
def classify_stock(quantity):
        if quantity == 0:
           return "Out of stock"
        elif (1<= quantity <= 10):
          return "Low stock"
        else:
          return "Well stocked"
for product in Products:
    print(product["name"],":", classify_stock(product["quantity"]))                
```

```python
def inventory_report(products):
    print("-----NKWEN TRADERS PRODUCT INVENTORY REPORT-----")
    for product in products:        
        print(f"Name: {product['name']}")
        print(f"Price: {product['price']}")
        print(f"Quantity: {product['quantity']}")
        print()
inventory_report(Products)    
```

```python
def calculate_inventory_value(products):
    Total = 0
    for product in products:        
        inventory_value = product['price'] * product['quantity']
        Total += inventory_value
    print(f"Total: {Total}FCFA")
calculate_inventory_value(Products)
```
