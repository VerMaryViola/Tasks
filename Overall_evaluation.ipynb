```python
# 40
class Inventory:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def total_inventory_value(self):
        total = 0
        for product in self.products:
            total += product.total_value()
        return total

inventory = Inventory()

inventory.add_product(Product("Rice", 5080, 7))
inventory.add_product(Product("Sugar", 2800, 4))
inventory.add_product(PerishableProduct("Milk", 250, 10, 5))
inventory.add_product(PerishableProduct("Bread", 1590, 5, 3))

print("Grand total:", inventory.total_inventory_value())
```

### Capstone Challenge 

```python

class Product:
    def __init__(self, name, price, quantity):
        self.name = name
        self.price = float(price)
        self.quantity = int(quantity)

    def total_value(self):
        return self.price * self.quantity

    def describe(self):
        return f"{self.name}:, quantity: {self.quantity}, Price: {self.price}"

class PerishableProduct(Product):
    def __init__(self, name, price, quantity, expiry_days):
        super().__init__(name, price, quantity)
        self.expiry_days = int(expiry_days)

    def describe(self):
        return f"{self.name}:, quantity: {self.quantity}, Price: {self.price}, Expiry_days{self.expiry_days})"

import csv
        # Reading CSV file with exception handling
def read_inventory(filename):
    products = []

    try:
        with open(filename, "r") as file:
            reader = csv.DictReader(file)
            for row in reader:
                if row["type"] == "perishable":
                    product = PerishableProduct(
                        row["name"],
                        row["price"],
                        row["quantity"],
                        row["expiry_days"]
                    )
                else:
                    product = Product(
                        row["name"],
                        row["price"],
                        row["quantity"]
                    )

                products.append(product)

    except FileNotFoundError:
        print(f"Error: This file was not found.")    
    return products

     # Function using *args
def report_line(*items):
    return " | ".join(str(item) for item in items)

      # Using filter() and lambda to find products needing reorder
def products_needing_reorder(products):
    return list(filter(lambda product: product.quantity < 10, products))

      # Using map() and lambda to get product names
def product_names(products):
    return list(map(lambda p: p.name, products))

def make_sales(products):
    sales = {
        "Rice": 5,
        "Milk": 3,
        "Sugar": 2
    }
    for product in products:
        if product.name in sales:
            product.quantity -= sales[product.name]


     # Updated inventory to CSV
def write_inventory(filename, products):
    with open(filename, "w") as file:        
        writer = csv.DictWriter(file, fieldnames = ["name", "price", "quantity", "type", "expiry_days"])
        writer.writeheader()

        for product in products:
            if isinstance(product, PerishableProduct):
                writer.writerow({
                    "name": product.name,
                    "price": product.price,
                    "quantity": product.quantity,
                    "type": "perishable",
                    "expiry_days": product.expiry_days
                })
            else:
                writer.writerow({
                    "name": product.name,
                    "price": product.price,
                    "quantity": product.quantity,
                    "type": "normal",
                    "expiry_days": ""
                })

        # The main program
def main():
    filename = "inventory.csv"
    products = read_inventory(filename)

    if not products:
        print("No inventory loaded.")
        return

    
    make_sales(products)        # Simulate a few sales    
    reorder = products_needing_reorder(products)    # Analyze inventory using filter/lambda   
    names = product_names(products)     # Analyze inventory using map/lambda

    # Calculating the total inventory value
    total_value = sum(map(lambda product: product.total_value(), products))

    # Finding the lowest-stock item
    lowest_stock = min(products, key=lambda product: product.quantity)

    # Fining the highest-value item
    highest_value = max(products, key=lambda product: product.total_value())

    # Updated inventory
    write_inventory("updated_inventory.csv", products)

    # 6. Final formatted report
    print("---------- NKWEN TRADERS STOCK MANAGER ----------")
    print(report_line("Products", len(products)))
    print(report_line("Product names", ", ".join(names)))
    print(report_line("Total inventory value", f"{total_value}"))
    print(report_line(
        "Lowest-stock item",
        f"{lowest_stock.name} ({lowest_stock.quantity} units)"
    ))
    print(report_line(
        "Highest-value item",
        f"{highest_value.name} ({highest_value.total_value()})"
    ))

    print("\nProducts needing reorder:")
    for product in reorder:
        print(f"- {product.name}: {product.quantity} units")
main()
```
