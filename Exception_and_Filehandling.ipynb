### Step 4:

```python
Products = [
    
     {'name': "Tomato Paste (tin)", 'price': 378.0, 'quantity': 16.0 },
     {'name': "Bread (loaf)", 'price': 591.0, 'quantity': 13.0},
     {'name': "Salt 1kg", 'price': 297.0, 'quantity': 11.0},
     {'name': "Rice 50kg", 'price': 34928.0, 'quantity': 1.0},
     {'name': "Detergent 1kg", 'price': 1768.0, 'quantity': 10.0}    
]
import csv

with open("Products.csv", "w", newline="") as file:
    fieldnames = ["name", "price", "quantity"]

    writer = csv.DictWriter(file, fieldnames=fieldnames)

    writer.writeheader()
    writer.writerows(Products)

print("CSV file created successfully.")
```

### Step 5:

#### - The main function of this code is to read the products stored in Products.csv and bring them back into Python as a list of dictionaries.
#### - The "DictReader" reads each row of the CSV and turns it into a dictionary and the "list(reader)" then collects all those rows and puts them into a Python list.

```python
with open("Products.csv", "r") as file:
    reader = csv.DictReader(file)

    products_from_file = list(reader)

print(products_from_file)
```

### Step 6:

```python
try:
    with open("Products.csv", "r") as file:
        reader = csv.DictReader(file)

        for row_number, row in enumerate(reader, start=2):
            try:
                name = row["name"]
               
                if row_number == 2:                        # Change the price of ROW 2 to an invalid value
                    row["price"] = "Five hundred"      

                price = float(row["price"])
                quantity = float(row["quantity"])

                print(f"{name} - Price: {price}, Quantity: {quantity}")

            except ValueError:
                print(f"Error in row {row_number}: Invalid price or quantity.")

except FileNotFoundError:
    print("Error: Products.csv was not found.")

finally:
    print("File-reading attempt complete.")
```

### Step 7:

#### - The except ValueError block caught the ValueError caused by trying to convert a non-numeric price such as "Five hundred" into a number. The except FileNotFoundError block caught the FileNotFoundError when the CSV filename was missing or misspelled. I know this because these were the error types Python showed before they were handled.

#### - If I had not used try/except for the missing-file case in Step 6, Python would have crashed the program and displayed a FileNotFoundError traceback instead of showing a friendly error message.
