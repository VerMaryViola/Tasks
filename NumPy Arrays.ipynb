# NumPy Arrays: Why Not Just Lists?

```python
import numpy as np
# Daily sales figures
daily_sales=[43750, 58200, 61300, 39900, 52400, 47600, 65100, 36200, 55800, 49300, 62500, 41500, 58700, 45200, 67100, 38500, 54300, 49700, 61000, 46900]
```

```python
# Converting the python list into Numpy array
sales_array = np.array(daily_sales)
print(sales_array)
```

```python
# Adding a 10% adjustment to all sales
adjusted_sales = sales_array * 1.10
print(adjusted_sales)
```

```python
# Calculating the mean sales
mean_sales = sales_array.mean()
print("Mean:", mean_sales)
```

```python
# Filtring values above the mean
above_mean = sales_array[sales_array > mean_sales]
print("Sales above mean:", above_mean)
```
