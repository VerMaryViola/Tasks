### Step 2:

##### This function calculates the total revenue for the whole week. This was done by multiplying each product's quantity by its price and summing them up together.

```python
def total_revenue(transactions):
    Total = 0
    for data in sales_data:
        Total += data["quantity"] * data["price"]
    return Total
    
print ("Total revenue : ", total_revenue(sales_data), "FCFA" )
```

### Step 3:

##### This functions dertemines the best selling product which is the product that had the highest total quantity sold during the week. A dictionary is used to keep track of the quantity sold for each product. From the result, Matches (box) is is the best with 55 as the number of quantity.
##### - An empty dictionary was created to store the total quantity sold and went through the data in the main list of dictionaries (sales_data) to perform the calculations.

```python
def best_selling_product(transactions):
    products = {}
    # Goes through every transaction in the sales_data and keeps the product, quantity  
    for data in sales_data:
        product = data["product"]        
        quantity = data["quantity"]
        
    # If the product is already in the dictionary, add the new quantity to its existing quantity.    
        if product in products:
            products[product] += quantity
        else:
            products[product] = quantity
        
    best_product = None
    best_quantity = 0
    for product in products:
        quantity = products[product]
        
        # Compares if the product quantity is greater than the best
        if quantity > best_quantity:
            best_product = product
            best_quantity = quantity
    
    return best_product, best_quantity
    
best_selling_product(sales_data)    
```

### Step 4:

##### This function organizes the transactions according to the day they occurred. For each day, this program calculates the total revenue for that day and prints the result accordingly.


```python
def day_by_day_breakdown(transactions):
    daily_sales = {}
    for sales in transactions:
        day = sales["day"]
        amnt = sales["quantity"] * sales["price"]
        
        if day in daily_sales:
            daily_sales[day] += amnt
        else:
            daily_sales[day] = amnt
    return daily_sales
    
daily_revenue = day_by_day_breakdown(sales_data)
for day, amnt in daily_revenue.items():
    print(day, "-", amnt, "FCFA")
```

### Step 5:

##### This function sorts the days according to thier revenue from the highest to the lowest. It was done using the "daily_revenue" which was gotten from the previous function.

```python
def sort_days_by_revenue(daily_revenue):
    # This dictionary keeps the unsorted days 
    remaining = {}     
    for day in daily_revenue:
        revenue = daily_revenue[day]
        remaining[day] = revenue
    # This keeps the sorted list    
    sorted= []
    # This compares the revenue with best if the lenght is greater then 0.
    while len(remaining) > 0:
        best_day = None
        best_revenue = 0
        for day in remaining:
            revenue = remaining[day]
            if revenue > best_revenue:
                best_day = day
                best_revenue = revenue  
                
        sorted.append((best_day, best_revenue))  # Appends the best day and revenue to the sorted list
        del remaining[best_day]    # Deletes the best day that has already been sorted and appended from the remaining dictionary
    return sorted
Sorted_list = sort_days_by_revenue(daily_revenue) 
for day, amnt in Sorted_list:
    print(day, "-", amnt, "FCFA")
```

### Step 6:

##### This function prints the combine results form all the other functions.

```python
def report(transactions):
    # Assigning variables to the other functions
    total = total_revenue(transactions)
    product, quantity = best_selling_product(transactions)
    daily_revenue = day_by_day_breakdown(transactions)
    ranked_days_by_revenue = sort_days_by_revenue(daily_revenue)
    
    print("------Traders Weekly Sales Report------")
    print()
    print(f"Weeks total revenue : {total:,} FCFA")
    print(f"Best selling_product : {product} ({quantity} units)")
    print()
    print("Days ranked by revenue(highest to lowest):")
    
    for day, revenue in ranked_days_by_revenue:
        print(f"  {day}: {revenue:,} FCFA")
    
```
