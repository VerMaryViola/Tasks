# The Empirical Rule & Z-Score

##### The 68-95-99.7 rule means that in a roughly normal distribution, about 68% of values are within 1 standard deviation of the mean,95% are within 2 standard deviations, and 99.7% are within 3 standard deviations.

## Step 2 - Learn Z-Scores
##### When we have a;
##### - z = 0 → the value is exactly at the mean.
##### - positive z-score → the value is above the mean.
##### - negative z-score → the value is below the mean.

```python
mean = 50000
std = 5000
sales = 60000
z = (sales - mean) / std
print(z)
```

##### From the result, it is a 2 standard deviation of the mean which is 95%.

## Step 3 - Connect the Empirical Rule to Z-Scores

##### A z-score of 2 is more unusual than a z-score of 0.5 because it is farther from the mean. A z-score of 0.5 is only half a standard deviation away, while a z-score of 2 is
two standard deviations away, making it much less common.

## Step 4 - Solve the Practice Problems

```python
mean = 50164.56
std = 4326.39
# List as used in my understanding distributions activity
daily_sales = [
    45000, 52000, 48000, 51000, 47000, 55000, 49000, 53000,59000, 46000, 50000, 
    54000, 48000,52000, 57000, 56000, 42514, 52000, 49000, 55000, 43500, 50000, 45000, 47000, 48100
]
for sales in daily_sales:
    z = (sales - mean) / std
    
    print(f"  Sales = {sales}")
    print(f"  Mean = {mean}")
    print(f"  Standard deviation = {std}")
    print(f"  Z-score = {z:.2f}")
    if abs(z) > 2:
        print("Statistical outlier")
    else:
        print("Normal variation")
    print("--------------------")
```

```python
|z| > 2

z = 0.5   → Normal variation
z = 1.2   → Normal variation
z = -1.8  → Normal variation
z = 2.1   → Outlier
z = -2.5  → Outlier
```
