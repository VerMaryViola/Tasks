```python
# 37
class Product:
    def __init__(self, name, price, quantity):        
        self.name = name
        self.price = price
        self.quantity = quantity

    def restock(self, amount):
        return self.quantity + amount 
        
    def total_value(self):
        return self.price * self.quantity 
P1 = Product("Matches", 2500, 5)
P2 = Product("Beans", 8930, 9)

print("- Updated quantity -")
print(P1.restock(23))
print(P2.restock(11))
```

```python
# 39
class Product:
    def __init__(self, name, price, quantity):        
        self.name = name
        self.price = price
        self.quantity = quantity
        

    def total_value(self):         
        return self.price * self.quantity

    def describe(self):
        return f"{self.name}: Price = {self.price}, Quantity = {self.quantity}"

class  PerishableProduct(Product):
    def __init__(self, name, price, quantity, expiry_days): 
        super().__init__(name, price, quantity)
        self.expiry_days = expiry_days    # adding attribute

    def describe(self):
        return f"{self.name}: Price = {self.price}, Quantity = {self.quantity}, Expiry dates = {self.expiry_days}"

products = [
    Product("Rice", 5000, 3),
    PerishableProduct("Milk", 250, 10, 5)
]

for product in products:
    print(product.describe())
    
```

```python
# 40
class Inventory:
    def __init__(self):
        self.products = []

    def add_product(self, product):
        self.products.append(product)

    def total_inventory_value(self):
        total = 0
        for product in self.products:
            total += product.total_value()
        return total

inventory = Inventory()

inventory.add_product(Product("Rice", 5080, 7))
inventory.add_product(Product("Sugar", 2800, 4))
inventory.add_product(PerishableProduct("Milk", 250, 10, 5))
inventory.add_product(PerishableProduct("Bread", 1590, 5, 3))

print("Grand total:", inventory.total_inventory_value())
```

### Capstone Challenge 

```python
import csv
class Product:          
    def __init__(self, name, quantity, price):
        self.name = name
        self.price = float(price)
        self.quantity = int(quantity)
    def total_value(self):
        return self.price * self.quantity
    def sell(self, amount):
        if amount <= self.quantity:
            self.quantity -= amount
        else:
            print(f"Not enough {self.name} in stock.")
    def describe(self):
        return f"{self.name}: {self.quantity}"

class PerishableProduct(Product):
    def __init__(self, name, quantity, price, expiry_days):
        super().__init__(name, quantity, price)
        self.expiry_days = expiry_days
    def describe(self):
        return (
            f"{self.name}: {self.quantity}"
            f"(expires in {self.expiry_days} days)"
        )

# The list of products written on the csv file
List_of_products = [
    {"name": "Book",  "quantity": 40, "price": 350},
    {"name": "pen", "quantity": 25, "price": 150},
    {"name": "School Bag", "quantity": 6, "price": 8000},
    {"name": "Pencil", "quantity": 50, "price": 100},
    {"name": "Calculator", "quantity": 8, "price": 5000}
]

with open("nkwen_products.csv", "w") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "quantity", "price"])
    writer.writeheader()
    writer.writerows(List_of_products)
products = []

try:
    with open("nkwen_products.csv", "r") as file:
        reader = csv.DictReader(file)

        for row in reader:
            product = Product(
                row["name"],
                row["quantity"],
                row["price"]                
            )
            products.append(product)
except FileNotFoundError:
    print("Error: Inventory file was not found.")

def print_report(*items):
    for item in items:
        print(item)
        
# Calling the sell method        
products[0].sell(15)
products[1].sell(27)
products[2].sell(16)
products[3].sell(15)

reorder_products = list(
    filter(lambda product: product.quantity < 10, products)
)
reorder_names = list(
    map(lambda product: product.name, reorder_products)
)

total_value = 0
for product in products:
    total_value += product.total_value()
lowest_stock = products[0]

for product in products:
    if product.quantity < lowest_stock.quantity:
        lowest_stock = product
highest_value = products[0]

for product in products:
    if product.total_value() > highest_value.total_value():
        highest_value = product

with open("nkwen_products_updated.csv", "w") as file:
    writer = csv.DictWriter(file, fieldnames=["name", "quantity", "price"])
    writer.writeheader()

    for product in products:
        writer.writerow({
            "name": product.name,
            "quantity": product.quantity,
            "price": product.price
        })
print()
print()
print_report(
    
    "----- NKWEN TRADERS STOCK REPORT -----",
    f"- Total inventory value: {total_value} FCFA",
    f"- Lowest-stock item: {lowest_stock.name} {lowest_stock.quantity}",
    f"- Highest-value item: {highest_value.name} ({highest_value.total_value()} FCFA)", 
    f"- Products needing reorder: {reorder_names}",
    "Updated inventory saved successfully."
)
```
