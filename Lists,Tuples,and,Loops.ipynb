## Activity 11: Lists, tuples and loops

```python
#A list of 10 Nkwen Traders daily sales totals
Sales_List = [34928.0, 8220.0, 3267.0, 7683.0, 5490.0, 35760.0	, 11666.0, 149656.0, 20910.0, 6248.0]

    
```

```python
# Calculating the total sales
Total_Sales = 0
for num in Sales_List:
    Total_Sales += num
print(Total_Sales)
```

```python
Avrg = Total_Sales/10
print(Avrg)
```

```python
Highest = Sales_List[0]
for num in Sales_List:
    if num > Highest:
        Highest = num
print(Highest)
```

```python
print("Answers")
print("Total: ",Total_Sales)
print(f"The average: {Avrg}")
print("The highest sales day: ", Highest)
```
