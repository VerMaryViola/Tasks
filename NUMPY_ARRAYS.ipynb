### Step 1 - Learn What Makes a NumPy Array Different from a List

##### Python list stores values as separate elements, while a NumPy array stores numerical data in a structured way that allows operations to be performed on all the elements at once.

```python
import numpy as np
# Creating Python list and Numpy array
numbers_list = [10, 20, 30, 40, 50]
numbers_array = np.array([10, 20, 30, 40, 50])
print("Python list:", numbers_list)
print("Numpy array:", numbers_array)
```

```python
# Multiplying each by two
print("List multiplied by 2:", numbers_list * 2)
print("Array multiplied by 2:", numbers_array * 2)
```

##### A Python list repeats its elements when multiplied by 2, while a NumPy array multiplies each element by 2. NumPy arrays are designed for numerical operations on the whole collection.

### Step 2 - Convert the Nkwen Traders Sales List to a NumPy Array

```python
sales = [ 
    58200, 43750, 61300, 39900, 52400, 47600, 65100, 36200, 55800, 49300,62000, 45500, 57100, 48900, 63500,41000, 54800, 59700, 46300, 68000
]
```

```python
sales_array = np.array(sales)

print("Original list:")
print(sales)

print("\nNumPy array:")
print(sales_array)
```

##### The Python list displays the values with commas, while the NumPy array displays them in an array format without commas. Both contain the same sales figures.

### Step 3 - Perform a Vectorized Adjustment (No Loop)

```python
adjusted_sales = sales_array * 1.10
print(adjusted_sales)
```

##### Every day’s sales figure is multiplied by 1.10 at once, which increases each sales value by 10%.

### Step 4 - Filter with Boolean Indexing (No Loop, No If-Statement)

```python
mean_sales = sales_array.mean()
print("Mean sales:", mean_sales)
```

```python
mask = sales_array > sales_array.mean()
print(mask)
```

```python
above_mean_sales = sales_array[mask]
print("Sales above the mean:", above_mean_sales)
```

### Step 5 - Write the Loop-Based Equivalent

```python
adjusted_sales_loop = []
for sale in sales:
    adjusted_sales_loop.append(sale * 1.10)
print(adjusted_sales_loop)
```

```python
above_mean_loop = []
for sale in sales:
    if sale > mean_sales:
        above_mean_loop.append(sale)
print("Sales above the mean:", above_mean_loop)
```

### Step 6 - Compare Line Count and Clarity

#A. Adjustment task
##### Vectorized NUmpy has 1 main code line.
##### Loop-based has 3 main code lines.

#B. Filtering task
##### Vectorized Numpy has 1 main code line.
##### Loop-based has 4 main code lines.

##### Therefore: Vectorized Numpy has a total of 2 lines which is shorter that Loop-based with total of 7 lines.

```python
import time
large_sales = np.random.randint(30000, 70000, 100000)  # This uses random numbers from the range of 30000 to 

# NumPy approach
start = time.time()
numpy_result = large_sales * 1.10
numpy_time = time.time() - start

# Loop-based approach
start = time.time()
loop_result = []
for sale in large_sales:
    loop_result.append(sale * 1.10)
loop_time = time.time() - start

print("NumPy time:", numpy_time)
print("Loop time:", loop_time)
```

### Comparison: 
##### The vectorized NumPy approach is shorter and easier to read at a glance. The loop-based approach requires more lines and makes the process of checking each individual value more explicit. NumPy hides the individual looping process but makes numerical operations more convenient and efficient.

### Final conclusion:
##### NumPy arrays allow numerical operations to be performed on many values at once. 
##### Vectorized operations and Boolean indexing make code shorter and easier to work with, while also being faster for large numerical datasets because the operations are optimized rather than relying on Python loops.
